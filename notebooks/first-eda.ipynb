# Bank Customer Churn — Orientation EDA

**Purpose:** a fast, read-only orientation to the 10,000-row Kaggle snapshot. This is not a substitute for the assessed Phase 1 preprocessing notebook; it simply verifies the raw schema before the KDD pipeline is run.

Run the assessed notebooks in order: Phase 1 → Phase 2 → Phase 3 → Phase 4 → Phase 5.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

PROJECT_ROOT = next(
    (
        candidate
        for candidate in (Path.cwd(), *Path.cwd().parents)
        if (candidate / 'data' / 'raw' / 'churn.csv').is_file()
    ),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError('Open this notebook from inside the Project_DATAMINING repository.')

RAW_PATH = PROJECT_ROOT / 'data' / 'raw' / 'churn.csv'
df = pd.read_csv(RAW_PATH)

print(f'Project root: {PROJECT_ROOT}')
print(f'Python: {sys.version.split()[0]}')
print(f'Raw shape: {df.shape[0]:,} rows × {df.shape[1]} columns')
display(df.head())
```

```python
orientation_audit = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'missing': df.isna().sum(),
    'unique': df.nunique(dropna=False),
})
display(orientation_audit)

print(f'Duplicate full rows: {df.duplicated().sum():,}')
print('Geography values:', sorted(df['Geography'].astype(str).unique()))
print('Gender values:', sorted(df['Gender'].astype(str).unique()))
print(f"Observed churn label rate: {df['Exited'].mean():.1%} (profiling lens, not a prediction target)")
```

## What this establishes

- The raw file contains 10,000 records and 14 columns, including three identifier/text fields that require an explicit keep/drop decision in Phase 1.
- `Exited` is retained only as a post-discovery profiling and validation lens.
- Column meanings such as `HasCrCard`, `IsActiveMember`, and `NumOfProducts` are interpreted as the dataset's binary/count fields. The source does **not** document a currency for `Balance` or `EstimatedSalary`, so later notebooks use currency-neutral “balance units.”
