There are 14 original raw features within 10,000 records, and `df.info()` confirms **no missing values in any column** (10,000 non-null everywhere) with dtypes already correct. The features group into five roles:

- **Identifiers:** `RowNumber`, `CustomerId`, `Surname` — record keys with no behavioral meaning; dropped during preprocessing (Surname is also PII).
- **Demographics:** `Geography` (France / Germany / Spain), `Gender`, `Age`.
- **Financial standing:** `CreditScore`, `Balance`, `EstimatedSalary`.
- **Relationship & behavior:** `Tenure` (years with the bank), `NumOfProducts` (1–4), `HasCrCard`, `IsActiveMember` (both 0/1).
- **Outcome — validation lens only:** `Exited` (1 = customer left the bank). Per the project brief this is used to profile and validate discovered patterns, not as a supervised prediction target.

**First observations from `head()` / `info()`:**

1. Two of the first five customers already show a balance of exactly **0.00** — an early hint of the zero-balance point-mass (~36% of all customers) that later shapes clustering and binning decisions.
2. Value scales differ by orders of magnitude (`NumOfProducts` 1–4 vs. `Balance`/`EstimatedSalary` ~10⁵), so any distance-based mining requires standardization first.
3. Three columns are text (`Surname`, `Geography`, `Gender`); only the last two carry analyzable categories.
